```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
def create_ages(mu=50,sigma=13,num_samples=100,seed=42):
    
    np.random.seed(42)

    sample_ages = np.random.normal(loc=mu,scale=13,size=num_samples)
    sample_ages =np.round(sample_ages,decimals=0)

    return sample_ages
```

```python
sample = create_ages()
```

```python
fig,ax=plt.subplots(nrows=1,ncols=2)
ax[0].hist(sample,bins=23);
ax[1].boxplot(sample);

```
