```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("Advertising.csv")
df.head()
```

```python
df['total_spent']=df['TV']+df['radio']+df['newspaper']
df.head()
```

```python
fig ,ax= plt.subplots(figsize=(10,5))
ax.scatter(x=df['total_spent'],y=df['sales'])

```

```python
X=df['total_spent']
y=df['sales']
```

```python
np.polyfit(X,y,deg=1)
```

```python
potential_spend=np.linspace(0,500,100)
```

```python
predicted_sales=0.04868788*potential_spend + 4.24302822
```

```python
fig, ax =plt.subplots()
ax.plot(potential_spend,predicted_sales, color="red")
ax.scatter(X,y)
ax.set(title="best line fit",
      xlabel="potential_spend",
      ylabel="Predicted_spend",
     )
```

```python
potential_spend.shape,predicted_sales.shape
```

```python
fig, (ax0,ax1) =plt.subplots(nrows=1,ncols=2,figsize=(10,5))
ax1.plot(potential_spend,predicted_sales, color="red")
ax1.scatter(X,y)
ax1.set(title="best line fit",
      xlabel="Total spend",
      ylabel="sales",
     )

ax0.scatter(X,y)
ax0.set(title="Advertising Plot",
      xlabel="Total spend",
      ylabel="Sales",
     )
```
